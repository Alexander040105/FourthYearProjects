```python
import pandas as pd
import numpy as np
import os 
import matplotlib
```

```python
current_dir = os.getcwd()
print(current_dir)

df = pd.read_csv(f'{current_dir}\\warehouse_messy_data.csv')
df.isnull().sum()
```

```python
df.dtypes
```

Cleaning the Quantity Column


```python
df.loc[df['Quantity'] == "two hundred", 'Quantity'] = '200'
df['Quantity'] = df['Quantity'].astype(float)
df['Quantity'] = df['Quantity'].fillna(df['Quantity'].mean())

```

```python
widgets = df['Product Name'].unique()
print(widgets)

average_price_per_widgets = df.groupby('Product Name')['Price'].mean().to_dict()
df['Price'] = df['Price'].fillna(df['Product Name'].map(average_price_per_widgets))
print(average_price_per_widgets)
```

```python
df.isnull().sum()
```

```python
df['Last Restocked'] = pd.to_datetime(df['Last Restocked'], errors='coerce')
```

```python
df = df.dropna()
print(df.isnull().sum())
print("\n")
print(f'Current number of data: {df.shape}')
print("\n")
print(df.dtypes)
```

```python
from sqlalchemy import create_engine

engine = create_engine('sqlite:///retail_warehouse.db')
df.to_sql('retail_warehouse', engine, if_exists='replace', index=False)
```
